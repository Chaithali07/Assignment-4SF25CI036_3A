**1. Counting Words in a File**

**Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.**

```python
def count_word(filename):
    with open(filename,'r') as file:
        text=file.read()
        words=text.split()
        return len(words)
print(count_word("C:\\Users\\CHAITHALI S\\Desktop\\SkillLab_Assignments\\Assignment\\document.txt"))
```

**2. Finding and Replacing Text**

**Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.**



```python
def fnd(old_word,new_word):
    with open('C:\\Users\\CHAITHALI S\\Desktop\\SkillLab_Assignments\\Assignment\\data.txt','r') as file:
        content=file.read()
    content=content.replace(old_word,new_word)
    with open('C:\\Users\\CHAITHALI S\\Desktop\\SkillLab_Assignments\\Assignment\\data.txt','w') as file:
        file.write(content)
    print("Text replaced successfully.")

fnd('Python','Java')
```

**3. Reading a File in Reverse**

**Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.**

```python
def reverse_file():
    with open('C:\\Users\\CHAITHALI S\\Desktop\\SkillLab_Assignments\\Assignment\\reverse.txt','r')as file:
        lines=file.readlines()
    for line in reversed(lines):
        print(line.strip())
reverse_file()
```

**15. Data Aggregation**

```python
df=pd.DataFrame({'Category':['A', 'C', 'B', 'C', 'B', 'A'],'Value':[12,35,56,43,65,76]})
df
```

```python
grouped_sum_mean=df.groupby('Category')['Value'].agg([np.sum,np.mean])
```
